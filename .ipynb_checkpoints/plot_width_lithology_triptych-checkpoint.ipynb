```python
import numpy as np
import matplotlib.pyplot as plt
import pyvista as pv
from landlab.io.native_landlab import load_grid
from llpvtools import grid_to_pv
```

```python
gridfw = load_grid("fixed_width3.grid")
```

```python
mfw, mfwc = grid_to_pv(gridfw, field_for_corner_z=0.0)
```

```python
gridfw.at_node.keys()
```

```python
cw = gridfw.at_node["channel_width"]
cw[cw == 0.0] = 1.0
logcw = np.log10(cw)
da = gridfw.at_node["drainage_area"] / 1.0e6
da[da == 0.0] = 1.0
logda = np.log10(da)
print(np.amax(logda))
```

```python
pl = pv.Plotter()
pl.set_scale(zscale=20)
_ = pl.add_mesh(mfw, cmap="GnBu", show_edges=True, scalars=logda, #"topographic__steepest_slope",
               scalar_bar_args=None, clim=[1.0, 4.0], line_width=0.5) #"channel_width")
pl.camera.position = (48545.119547239796, -186171.23484915978, 203855.18282957753)
pl.show()
```

```python
def plotsa(grid, psym="k."):
    slp = grid.at_node["topographic__steepest_slope"][grid.core_nodes]
    area = grid.at_node["drainage_area"][grid.core_nodes] / 1.0e6
    plt.loglog(area, slp, psym)
    plt.xlabel(r"Drainage area (km$^2$)", fontsize=14)
    plt.ylabel("Topographic gradient (m/m)", fontsize=14)
    plt.grid(True)
```

```python
plotsa(gridfw, psym='b.')
plotsa(gridnt, psym='r+')
#plotsa(gridntml, psym='cx')
plt.legend(['Fixed width', 'Near-threshold width'])
plt.savefig("slope-area.pdf")
```

```python
def plotwid(grid, psym="k.", ax=None):
    wid = grid.at_node["channel_width"][grid.core_nodes]
    area = grid.at_node["drainage_area"][grid.core_nodes] / 1.0e6
    if ax is None:
        plt.loglog(area, wid, psym)
        plt.xlabel(r"Drainage area (km$^2$)", fontsize=14)
        plt.ylabel("Channel width (m)", fontsize=14)
        plt.grid(True)
    else:
        ax.loglog(area, wid, psym)
        ax.set_xlabel(r"Drainage area (km$^2$)")
        ax.set_ylabel("Channel width (m)")
        ax.grid(True)
        
```

```python
plotwid(gridfw, psym='b.')
plotwid(gridnt, psym='r+')
plt.legend(['Fixed width', 'Near-threshold width'])
plt.savefig("width-area.pdf")
#plotwid(gridntml, psym='cx')
```

```python
def get_area_slp_wid(grid):
    a = grid.at_node["drainage_area"][grid.core_nodes] / 1.0e6
    s = grid.at_node["topographic__steepest_slope"][grid.core_nodes]
    w = grid.at_node["channel_width"][grid.core_nodes]
    return a, s, w

afw, sfw, wfw = get_area_slp_wid(gridfw)
ant, snt, wnt = get_area_slp_wid(gridnt)
```

```python
fig, axs = plt.subplot_mosaic(
    [['(a)', '(a)'], ['(b)', '(b)'], ['(c)', '(d)']], 
    gridspec_kw={'height_ratios': [1.5, 1.5, 1]},
    figsize=(6.5, 9)
)
fig.tight_layout()

axs['(a)'].imshow(fwimg)
axs['(a)'].axis(False)
axs['(b)'].imshow(ntimg)
axs['(b)'].axis(False)

axs['(a)'].plot([400, 2250], [1060, 710], 'k', linewidth=1)
axs['(a)'].text(1300, 950, '160 km')

axs['(c)'].loglog(afw, sfw, 'b.')
axs['(c)'].loglog(ant, snt, 'rx')
axs['(c)'].set_xlabel(r"Drainage area (km$^2$)")
axs['(c)'].set_ylabel("Topographic gradient (m/m)")
axs['(c)'].grid(True)
axs['(c)'].legend(['Fixed width', 'Near threshold'],
                 fontsize=9)

axs['(d)'].loglog(afw, wfw, 'b.')
axs['(d)'].loglog(ant, wnt, 'rx')
axs['(d)'].set_xlabel(r"Drainage area (km$^2$)")
axs['(d)'].set_ylabel("Channel width (m)")
axs['(d)'].grid(True)

for label, ax in axs.items():
    # Use Axes.annotate to put the label
    # - at the top left corner (axes fraction (0, 1)),
    # - offset half-a-fontsize right and half-a-fontsize down
    #   (offset fontsize (+0.5, -0.5)),
    # i.e. just inside the axes.
    if label=='(c)':
        xyt = (+3.5, -0.5)
    else:
        xyt = (+0.5, -0.5)
    ax.annotate(
            label,
            xy=(0, 1), xycoords='axes fraction',
            xytext=xyt, textcoords='offset fontsize',
            fontsize='medium', verticalalignment='top', fontfamily='serif',
            #bbox=dict(facecolor='0.7', edgecolor='none', pad=3.0)
        )

plt.tight_layout()

plt.savefig("width_comparison.pdf")
```
